## Day 2 ⚛️ Entangle two qubits

`H` puts qubit 0 into superposition. Then a CNOT (`cx`) links qubit 1 to it: it flips qubit 1 exactly when qubit 0 is 1. Because qubit 0 is a mix of 0 and 1, the pair ends up as a mix of "both 0" and "both 1". That's a **Bell pair**: two qubits that share one fate.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)                      # qubit 0: superposition
qc.cx(0, 1)                  # CNOT: flips qubit 1 when qubit 0 is 1. This entangles them.
qc.measure([0, 1], [0, 1])   # measure both
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("Entangled pair:", counts)
```

You should see only `00` and `11`, roughly half each. Never `01`, never `10`. Each qubit on its own is a fair coin, yet the two always agree.

**So can you use this to send a message faster than light?** No. Each side, alone, just sees random 0s and 1s. The agreement only shows up when you compare notes, and comparing notes needs an ordinary message.

In 2022 the Nobel Prize in Physics went to Alain Aspect, John Clauser and Anton Zeilinger for experiments with entangled photons. You just made an entangled pair of your own (on a simulator) in a few lines of code.

## Day 4 ⚛️ Make two H gates cancel

One `H` gives you 50/50. So two `H` gates in a row should give 50/50 too, like flipping a coin twice. Right? Run it and see.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0)            # into superposition...
qc.h(0)            # ...and straight back out again
qc.measure(0, 0)
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("Two H gates:", counts)
```

You should see `{'0': 1000}`. Every single shot.

Here's why. After the first `H`, the qubit has two **amplitudes**, one for 0 and one for 1, both +1/√2. The second `H` splits each of them again. The two paths that lead to 0 have the same sign, so they add up. The two paths that lead to 1 have opposite signs, so they cancel to zero. That's **interference**: amplitudes (not probabilities) adding and cancelling. Coins can't do it. It's the engine behind quantum algorithms, starting with the Grover search you'll build in Ladder 2.

## Day 6 ⚛️ Three qubits that always agree (GHZ)

Take your Bell pair and add a third qubit to the chain.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 3)
qc.h(0)                            # qubit 0: superposition
qc.cx(0, 1)                        # entangle qubit 1 with qubit 0
qc.cx(1, 2)                        # ...and qubit 2 with qubit 1
qc.measure([0, 1, 2], [0, 1, 2])
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("GHZ state:", counts)
```

You should see only `000` and `111`, roughly half each. Three qubits, one shared fate: measure any one of them and you know the other two.

This is a **GHZ state**, named after the physicists Daniel Greenberger, Michael Horne and Anton Zeilinger, who first wrote about states like it in 1989. Zeilinger's group made a three-photon GHZ state in the lab in 1999, and he went on to share that 2022 Nobel Prize. You just made one in a few lines of code. Real quantum hardware comes on Day 30.
