## Day 2 · Entangle two qubits

`H` puts qubit 0 into superposition. A CNOT gate (`cx`) then links qubit 1 to it: it flips qubit 1 exactly when qubit 0 is 1. Because qubit 0 is a combination of 0 and 1, the pair ends up as a combination of "both 0" and "both 1". This is a **Bell pair**: two qubits that share one outcome.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)                      # qubit 0: superposition
qc.cx(0, 1)                  # CNOT: flips qubit 1 when qubit 0 is 1. This entangles them.
qc.measure([0, 1], [0, 1])   # measure both
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("Entangled pair:", counts)
```

> **What you should see:** only `00` and `11`, roughly half each. Never `01`, never `10`.

Each qubit on its own behaves like a fair coin, yet the two always agree.

**Can this send a message faster than light?** No. Each side, on its own, sees only random 0s and 1s. The agreement appears only when the two sides compare results, and comparing results needs an ordinary, classical message.

In 2022 the Nobel Prize in Physics went to Alain Aspect, John Clauser and Anton Zeilinger for experiments with entangled photons. You have just made an entangled pair of your own, on a simulator, in a few lines of code.

**Try this.** Change `shots=1000` to `shots=10` and run the cell a few times. You will still never see `01` or `10`.

## Day 4 · Make two H gates cancel

One `H` gate gives a 50/50 result. So two `H` gates in a row should also give 50/50, like flipping a coin twice. Run the cell and see.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0)            # into superposition...
qc.h(0)            # ...and straight back out again
qc.measure(0, 0)
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("Two H gates:", counts)
```

> **What you should see:** `{'0': 1000}`, on every shot.

**Why.** After the first `H`, the qubit has two **amplitudes**, one for 0 and one for 1, both +1/√2. The second `H` splits each of them again. The two paths that lead to 0 have the same sign, so they add up. The two paths that lead to 1 have opposite signs, so they cancel to zero. This is **interference**: amplitudes, not probabilities, adding and cancelling. A coin cannot do this, and it is the engine behind quantum algorithms, starting with the Grover search you build in Project Ladder 2.

**Puzzle.** Add one gate between the two `H` gates so the result becomes `1` on every shot. Hint: you want a gate that flips the sign of the 1 part and leaves the 0 part alone.

<details>
<summary><b>Show the answer</b></summary>

Add `qc.z(0)` between the two `H` gates. The `Z` gate flips the sign of the 1 amplitude, so now the paths to 0 cancel and the paths to 1 add up. In the language of gates, `H Z H` is the same as `X`, the quantum NOT gate.

## Day 6 · Three qubits that always agree

Take your Bell pair and add a third qubit to the chain.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 3)
qc.h(0)                            # qubit 0: superposition
qc.cx(0, 1)                        # entangle qubit 1 with qubit 0
qc.cx(1, 2)                        # ...and qubit 2 with qubit 1
qc.measure([0, 1, 2], [0, 1, 2])
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("GHZ state:", counts)
```

> **What you should see:** only `000` and `111`, roughly half each.

Three qubits, one shared outcome: measure any one of them and you know the other two.

This is a **GHZ state**, named after the physicists Daniel Greenberger, Michael Horne and Anton Zeilinger, who first described states like it in 1989. Zeilinger's group created a three-photon GHZ state in the laboratory in 1999, and he went on to share the 2022 Nobel Prize.

**Try this.** Extend it to four qubits: use `QuantumCircuit(4, 4)`, add `qc.cx(2, 3)`, and measure all four.
