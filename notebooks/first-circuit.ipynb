## Step 2 · Build your circuit and run it 1,000 times

The circuit has one qubit and three instructions:

1. A **Hadamard gate** (`H`) puts the qubit into **superposition**: a combination of 0 and 1 that settles on one of them only when it is measured.
2. A **measurement** reads the qubit and stores the answer in a classical bit.
3. The **simulator** runs the whole circuit 1,000 times (1,000 *shots*) and counts the answers.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)   # 1 qubit, plus 1 classical bit to hold its answer
qc.h(0)                     # Hadamard gate: superposition
qc.measure(0, 0)            # measure qubit 0, write the answer into bit 0
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("My first quantum results:", counts)
```

> **What you should see:** your circuit drawn in text (the `H` box, then the measurement `M`), followed by your counts, for example `{'0': 496, '1': 504}`.

Your numbers will differ from these, and they change every time you run the cell. That is not a bug: it is sampling, and it is exactly what should happen. The last section explains why.

## Step 3 · Chart the results

```python
from qiskit.visualization import plot_histogram

plot_histogram(counts, title="My first quantum circuit: 1,000 shots")
```

## You've run a quantum circuit

That is real Qiskit code: the same toolkit the Day 30 course uses to send circuits to a real IBM quantum computer.

### Why not exactly 500 and 500?

Every shot is a fresh measurement with a 50% chance each way, so the totals vary. With 1,000 shots, about 95 runs in 100 land between 469 and 531 zeros, and exactly 500 turns up only about once in 40 runs. As a share of all your shots, the variation shrinks as you add more: four times the shots halves it. This is how quantum engineers decide how many shots a job needs.

### What the simulator is doing

This simulator is classical software. It calculates the 50/50 split exactly, then picks each outcome with a pseudo-random number generator. On a real quantum computer the randomness is physical. You will run on real hardware when your Day 30 course opens.
