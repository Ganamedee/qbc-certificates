## Step 2: build your circuit and run it 1,000 times

One qubit. One Hadamard gate (`H`) to put it into **superposition**: a mix of 0 and 1 that only picks one when you measure it. Then one measurement. Then the simulator runs the whole circuit 1,000 times and counts the answers.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)   # 1 qubit, plus 1 classical bit to hold its answer
qc.h(0)                     # Hadamard gate: superposition
qc.measure(0, 0)            # measure qubit 0, write the answer into bit 0
print(qc.draw())

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print("My first quantum results:", counts)
```

You should see your circuit drawn in text (the `H` box, then the meter `M`) and your counts, something like `{'0': 496, '1': 504}`.

Your numbers will be different from mine. Run the cell again and they change again. That's not a bug. It's sampling, and it's exactly what should happen (more on that at the bottom).

## Step 3: see it

```python
from qiskit.visualization import plot_histogram

plot_histogram(counts, title="My first quantum circuit: 1,000 shots")
```

## 🎉 You just ran a quantum circuit

That's real Qiskit code: the same toolkit the 🎁 Day 30 Unlock uses to send circuits to a real IBM quantum computer.

**Why not exactly 500/500?** Every shot is a fresh measurement with a 50% chance each way, so the totals wobble. With 1,000 shots, about 95 runs in 100 land between 469 and 531 zeros, and exactly 500 turns up only about once in 40 runs. Measured as a share of all your shots, the wobble shrinks as you add more (four times the shots halves it), which is exactly how quantum engineers decide how many shots a job needs.

**One honest note:** this simulator is classical software. It works out the 50/50 exactly, then picks each outcome with a pseudo-random number generator. On a real quantum computer the randomness is physical, and your real-hardware course unlocks on Day 30.
